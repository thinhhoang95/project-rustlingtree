# HLLRD Build Matrix

Build one cluster-specific aligned normal-residual matrix from raw ADS-B tracks.

```python
from pathlib import Path
from hllrd.cli import build_matrix_main

output = Path('../../data/hllrd/se_matrix.npz')
build_matrix_main([
    '--cluster', 'SE',
    '--raw-adsb-dir', '../../data/adsb/raw',
    '--cluster-artifacts', '../../data/artifacts/simap_arrival_flights.jsonl',
    '--stations', '200',
    '--output', str(output),
])
```

```python
from hllrd.matrix import load_matrix_artifact

matrix = load_matrix_artifact(output)
matrix.X_centered.shape, matrix.cluster, matrix.sigma_hat, matrix.flight_ids[:5]
```
