# HLLRD Transform Test Matrix

Apply a learned event dictionary to a held-out matrix with the same station layout.

```python
from pathlib import Path
from hllrd.cli import transform_main

matrix_path = Path('../../data/hllrd/se_matrix.npz')
model_path = Path('../../data/hllrd/se_model.npz')
transform_path = Path('../../data/hllrd/se_transform.npz')
transform_main([
    '--matrix', str(matrix_path),
    '--model', str(model_path),
    '--output', str(transform_path),
    '--summary', '../../data/hllrd/se_transform_summary.json',
])
```

```python
import numpy as np

with np.load(transform_path) as data:
    float(data['explained_fraction']), data['active_counts'][:10]
```
