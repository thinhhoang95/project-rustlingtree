# HLLRD Fit Model

Fit the V1 fixed-rank localized low-rank model on a cluster matrix.

```python
from pathlib import Path
from hllrd.cli import fit_main

matrix_path = Path('../../data/hllrd/se_matrix.npz')
model_path = Path('../../data/hllrd/se_model.npz')
fit_main([
    '--matrix', str(matrix_path),
    '--output', str(model_path),
    '--summary', '../../data/hllrd/se_summary.json',
    '--events-csv', '../../data/hllrd/se_events.csv',
    '--energy-plot', '../../data/hllrd/se_energy.png',
    '--heatmap', '../../data/hllrd/se_heatmap.png',
])
```

```python
from hllrd.fit import load_fit_result, event_summary

result = load_fit_result(model_path)
result.explained_fraction, event_summary(result)[:5]
```
